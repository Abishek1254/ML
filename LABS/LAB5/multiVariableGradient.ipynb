```python
import torch
a=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(1.0,requires_grad=True)

g= a**2 + b**2 + a*b
print("Value of function f : ",g.item())

g.backward()
# how much f changes if x changes
print("Gradient dg/da: ",a.grad)
# how much f changes if y changes
print("Gradient dg/db ",b.grad)
```

```python
l=torch.tensor(-2.0,requires_grad=True)
m=torch.tensor(4.0,requires_grad=True)

h=l**2 + m**2 + l*m
print("Function value : ",h.item())

h.backward()

print("Gradient dh/dl : ",l.grad)
print("Gradient dh/dm : ",m.grad)
```
