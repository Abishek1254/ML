```python
# -----------------------------
# Step 1: Initialize
# -----------------------------
x = 2.0              # input
y_true = 10.0        # actual value we want

W = 1.0              # initial weight (random guess)
b = 0.0              # initial bias (random guess)

# optimal parameters should be w=4 and b=2
# lr- learning rate
lr = 0.01

epochs=[10,20,30,40,50]
```

```python
# -----------------------------
# Step 2: Training loop
# -----------------------------
for i in range(epochs[0]):
    # ---- Forward pass ----
    y_pred = W * x + b

    # ---- Loss (Mean Squared Error with 1/2 factor) ----
    loss = 0.5 * (y_pred - y_true)**2

    # ---- Gradients (manual) ----
    dL_dW = (y_pred - y_true) * x
    dL_db = (y_pred - y_true)

    # ---- Update ----
    W = W - lr * dL_dW
    b = b - lr * dL_db

    # ---- Print progress ----
    print(f"Iteration {i+1}: Loss = {loss:.4f}, W = {W:.4f}, b = {b:.4f}")
```
