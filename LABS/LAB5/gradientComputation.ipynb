```python
import torch
# define tensor x with gradient tracking
# You create a tensor with value 2.0 , requires_grad=True tells PyTorch:
# “Track all operations on this tensor because I will need its gradient later.”
x=torch.tensor(2.0,requires_grad=True)
f= x**2+3*x+5

# this tells pytorch - 
# “Compute derivative of function f  w.r.t all tensors that have requires_grad=True”
# PyTorch applies chain rule backward through the graph.
# backward() works directly only when: op is a scalar
f.backward()

# f is a tensor
# .item() converts it to a normal Python number
print("Function value:", f.item())
# x.grad stores the derivative df/dx
print("Gradient (df/dx):", x.grad)
```

```python
y=torch.tensor(3.0,requires_grad=True)
f1= y**2+3*y
print("Function value : ",f1.item())
f1.backward()
print("Gradient df1/dy: ",y.grad)
```
