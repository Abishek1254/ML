```python
# -----------------------------
# Step 1: Initialize
# -----------------------------
x = 2.0          # input
y_true = 10.0    # target

# Layer 1 (input → hidden)
W1 = 1.0
b1 = 0.0

# Layer 2 (hidden → output)
W2 = 1.0
b2 = 0.0

lr = 0.01
epochs = 20
```

```python
# -----------------------------
# ReLU function and derivative
# -----------------------------
def relu(z):
    return max(0, z)

def relu_derivative(z):
    if(z>0):
        return 1.0
    else:
        return 0.0
    

# -----------------------------
# Training loop
# -----------------------------
for i in range(epochs):
    # ---- Forward pass ----
    z1 = W1 * x + b1          # hidden linear
    h = relu(z1)              # activation
    y_pred = W2 * h + b2      # output

    # ---- Loss ----
    loss = 0.5 * (y_pred - y_true)**2

    # ---- Backprop ----
    error = (y_pred - y_true)

    # Output layer gradients
    dL_dW2 = error * h
    dL_db2 = error

    # Hidden layer gradients
    dL_dh = error * W2
    dL_dz1 = dL_dh * relu_derivative(z1)

    dL_dW1 = dL_dz1 * x
    dL_db1 = dL_dz1

    # ---- Update ----
    W2 = W2 - lr * dL_dW2
    b2 = b2 - lr * dL_db2

    W1 = W1 - lr * dL_dW1
    b1 = b1 - lr * dL_db1

    # ---- Print progress ----
    print(f"Iter {i+1}: Loss={loss:.4f}, y_pred={y_pred:.4f}, W1={W1:.4f}, W2={W2:.4f}")
```
