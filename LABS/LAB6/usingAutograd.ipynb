```python
# Overall Flow -
# Tell PyTorch which variables need gradients
# Do forward pass normally
# Call .backward()
# PyTorch fills .grad automatically

import torch
# -----------------------------
# Step 1: Initialize
# -----------------------------
x = torch.tensor(2.0)
y_true = torch.tensor(10.0)

# Parameters (with gradient tracking)
W1 = torch.tensor(1.0, requires_grad=True)
b1 = torch.tensor(0.0, requires_grad=True)

W2 = torch.tensor(1.0, requires_grad=True)
b2 = torch.tensor(0.0, requires_grad=True)

lr = 0.01
epochs = 20
```

```python
# -----------------------------
# Training loop
# -----------------------------
for i in range(epochs):
    # ---- Forward pass ----
    z1 = W1 * x + b1
    h = torch.relu(z1)
    y_pred = W2 * h + b2

    # ---- Loss ----
    loss = 0.5 * (y_pred - y_true)**2

    # ---- Backward pass ----
    loss.backward()

    # ---- Update ----
    with torch.no_grad():   # stop tracking updates
        W2 -= lr * W2.grad
        b2 -= lr * b2.grad

        W1 -= lr * W1.grad
        b1 -= lr * b1.grad

    # ---- Zero gradients ----
    # PyTorch accumulates gradients. So before next iteration: reset gradients to zero
    # Otherwise gradients will stack up incorrectly.
    W1.grad.zero_()
    b1.grad.zero_()
    W2.grad.zero_()
    b2.grad.zero_()

    # ---- Print progress ----
    print(f"Iter {i+1}: Loss={loss.item():.4f}, y_pred={y_pred.item():.4f}")
```
