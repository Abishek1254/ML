```python
import torch
# 2D empty tensor
x=torch.empty(2,2)
print(x)

# 3D empty tensor
x=torch.empty(2,2,2)
print(x)
```

```python
x=torch.rand(3,3)
print(x)

y=torch.zeros(2,2,2)
print(y)

x=torch.ones(2,2,3)
print(x.size())

# tensor creation using python list
y=torch.tensor([3,5,6,7,9])
print(y)
```

```python
# Basic operations on tensors
x=torch.tensor([2,2,3])
y=torch.tensor([2,6,6])

# Addition
z1=x+y
z2=torch.add(x,y)
print("z1 : ",z1)
print("z2 : ",z2)

# Inplace addition
y.add_(x)
print("y : ",y)
```

```python
# Subtraction
x=torch.tensor([2,2,3])
y=torch.tensor([2,6,6])

z1=x-y
z2=torch.sub(x,y)

print("z1 : ",z1)
print("z2 : ",z2)
```

```python
# Multiplication
x=torch.tensor([2,2,3])
y=torch.tensor([2,6,6])

z1=x*y
z2=torch.mul(x,y)

print("z1 : ",z1)
print("z2 : ",z2)
```
